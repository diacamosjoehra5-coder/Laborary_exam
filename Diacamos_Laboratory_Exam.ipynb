```python
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="total day minutes",
    y="total day charge",
    hue="churn",
    style="churn",
    s=100
)

plt.title("Total Day Minutes vs. Total Day Charge by Churn Status")
plt.xlabel("Total Day Minutes")
plt.ylabel("Total Day Charge")
plt.legend(title="Churned")
plt.show()
```

```python
# Calculate the mean value of numeric metrics grouped by churn status
numeric_cols = df.select_dtypes(include=['number']).columns.tolist() + ['churn']
grouped_metrics = df[numeric_cols].groupby('churn').mean()

print("Average metrics by Churn status:")
display(grouped_metrics)
```
